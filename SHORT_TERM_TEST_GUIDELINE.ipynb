# Puppy App Short-Term Test Guideline

Use this notebook to run a compressed local trial test of the Puppy App in VS Code.

**Current local setting:** `backend/.env` contains `PUPPY_STUDY_DAY_MINUTES=60`, so **1 study day = 60 real minutes** in local PostgreSQL mode.

## What this test is for

- Verify that a participant can complete Day 1 and unlock Day 2 one hour later.
- Verify that repeat submissions are blocked within the same study block.
- Verify that later study days load correctly.
- Verify that recovery code restore still works.

## Before you start

You need:

- Docker or Rancher Desktop running
- Local PostgreSQL container available
- Node backend started on port `3001`
- An active trial order in `/admin`
- A private/incognito browser window
- A place to save the participant recovery code

## Step 1 — Start the local environment

Run these commands from the project root in the VS Code terminal:

```powershell
docker compose up -d
docker compose ps
npm run dev
```

Then check:

- `http://localhost:3001/api/health`
- `http://localhost:3001/admin`

Confirm the health endpoint reports the database is connected, and confirm a trial-order set is active in admin.

## Step 2 — Start a participant test

Open a private/incognito browser window and go to:

`http://localhost:3001/?backend=postgres`

Then:

1. Create a new synthetic participant.
2. Save the recovery code shown by the app.
3. Complete the full visit for the current study day.

## Step 3 — Verify the lock behavior

Immediately after completion:

- Refresh the participant page
- Confirm the app shows the locked/completed state
- Confirm you cannot submit another response during the same 60-minute block

If you want to repeat Day 1 immediately, create a **new participant** in a fresh incognito window.

## Step 4 — Verify next-day unlock

Wait until **60 minutes have passed since enrollment** for that participant. Then:

1. Reopen the participant app, or restore using the saved recovery code.
2. Confirm the next study day is available.
3. Complete the visit and confirm it saves successfully.

In this quick-test mode, the study day advances by **elapsed time since enrollment**, not by midnight.

## Suggested one-day test plan

| Time | Action | Verify |
| --- | --- | --- |
| Hour 0 | Create participant and complete Day 1 | Response saves and participant locks |
| Hour 1 | Return and complete Day 2 | New trial loads and save succeeds |
| Hour 2 | Restore participant with recovery code | Recovery flow works |
| Hour 3+ | Repeat with same participant | Each new block unlocks one new study day |

## Optional database check

In a second terminal, run:

```powershell
npm run db:studio
```

Then inspect `Participant` and `DailyResponse` to confirm the expected `studyDay` rows were saved.

## If you want it faster

Edit `backend/.env` and change:

```env
PUPPY_STUDY_DAY_MINUTES=60
```

Examples:

- `30` → one study day every 30 minutes
- `15` → one study day every 15 minutes
- `1440` → normal one-day schedule

After changing it, restart the backend with `npm run dev`.

## Quick checklist

- [ ] Database starts successfully
- [ ] Backend health check passes
- [ ] Trial order is active in admin
- [ ] Day 1 can be completed
- [ ] Same-day repeat is blocked
- [ ] Next study day unlocks after 60 minutes
- [ ] Recovery code restore works
